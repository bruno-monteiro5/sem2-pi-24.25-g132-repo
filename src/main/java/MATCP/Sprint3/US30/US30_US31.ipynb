```
{
 "cell_type": "markdown",
 "metadata": {},
 "source": [
    "# User Story 30 - Algorithm Efficiency Analysis\n",
    "\n",
    "## Description 📄\n",
    "As a Product Owner, I want to analyze the efficiency of the algorithms developed in US13 and US14 by obtaining the best-fitting curve for execution time samples collected in US29. The fitted curve should align with the theoretical complexity analysis from US26/US28, with graphical visualization of the results.\n",
    "\n",
    "---\n",
    "\n",
    "## 1. Objective 🎯\n",
    "\n",
    "This analysis aims to:\n",
    "1. Determine the empirical time complexity of US13 (connectivity verification) and US14 (maintenance route) algorithms\n",
    "2. Compare with expected theoretical complexity\n",
    "3. Identify potential optimization opportunities through curve fitting\n",
    "\n",
    "---\n",
    "\n",
    "## 2. Theoretical Background 📚\n",
    "\n",
    "### Complexity Models Tested:\n",
    "\n",
    "- **Linear** ($O(n)$):\n",
    "$$ T(n) = an + b $$\n",
    "\n",
    "- **Quadratic** ($O(n^2)$):\n",
    "$$ T(n) = an^2 + bn + c $$\n",
    "\n",
    "- **Linearithmic** ($O(n\\log n)$):\n",
    "$$ T(n) = an\\log n + b $$\n",
    "\n",
    "### Evaluation Metric:\n",
    "**Root Mean Square Error (RMSE):**\n",
    "$$ RMSE = \\sqrt{\\frac{1}{n}\\sum_{i=1}^n(y_i-\\hat{y}_i)^2} $$\n",
    "\n",
    "### Theoretical Expectations:\n",
    "- US13: Should exhibit $O(V+E)$ linear complexity\n",
    "- US14: Should exhibit $O(E)$ linear complexity\n",
    "\n",
    "---\n",
    "\n",
    "## 3. Code 💻"
 ]
}
```

```
{
 "cell_type": "code",
 "execution_count": null,
 "metadata": {},
 "outputs": [],
 "source": [
    "import pandas as pd\n",
    "import numpy as np\n",
    "import matplotlib.pyplot as plt\n",
    "from scipy.optimize import curve_fit\n",
    "\n",
    "# Load and prepare data\n",
    "df = pd.read_csv(\"data_us29.csv\")\n",
    "df['tempo_us13_ms'] = df['tempo_us13'] * 1000\n",
    "df['tempo_us14_ms'] = df['tempo_us14'] * 1000\n",
    "\n",
    "# Define complexity models\n",
    "def linear(x, a, b): return a*x + b\n",
    "def quadratic(x, a, b, c): return a*x**2 + b*x + c\n",
    "def nlogn(x, a, b): return a*x*np.log(x) + b\n",
    "\n",
    "# US13 Analysis\n",
    "print(\"\\n🔍 US13 (Connectivity Verification) Analysis\")\n",
    "params_us13_lin, _ = curve_fit(linear, df['input_size'], df['tempo_us13_ms'])\n",
    "params_us13_quad, _ = curve_fit(quadratic, df['input_size'], df['tempo_us13_ms'])\n",
    "params_us13_nlogn, _ = curve_fit(nlogn, df['input_size'], df['tempo_us13_ms'])\n",
    "\n",
    "# US14 Analysis\n",
    "print(\"\\n🔍 US14 (Maintenance Route) Analysis\")\n",
    "params_us14_lin, _ = curve_fit(linear, df['input_size'], df['tempo_us14_ms'])\n",
    "params_us14_quad, _ = curve_fit(quadratic, df['input_size'], df['tempo_us14_ms'])\n",
    "\n",
    "# Visualization\n",
    "plt.figure(figsize=(14,6))\n",
    "x_range = np.linspace(min(df['input_size']), max(df['input_size']), 100)\n",
    "\n",
    "# US13 Plot\n",
    "plt.subplot(1,2,1)\n",
    "plt.scatter(df['input_size'], df['tempo_us13_ms'], label='Empirical Data')\n",
    "plt.plot(x_range, quadratic(x_range, *params_us13_quad), 'r-', label='Quadratic Fit')\n",
    "plt.plot(x_range, nlogn(x_range, *params_us13_nlogn), 'g--', label='O(n log n) Fit')\n",
    "plt.title('US13 - Connectivity Verification')\n",
    "plt.xlabel('Input Size (n)')\n",
    "plt.ylabel('Execution Time (ms)')\n",
    "plt.legend()\n",
    "\n",
    "# US14 Plot\n",
    "plt.subplot(1,2,2)\n",
    "plt.scatter(df['input_size'], df['tempo_us14_ms'], label='Empirical Data')\n",
    "plt.plot(x_range, linear(x_range, *params_us14_lin), 'b-', label='Linear Fit')\n",
    "plt.title('US14 - Maintenance Route')\n",
    "plt.xlabel('Input Size (n)')\n",
    "plt.ylabel('Execution Time (ms)')\n",
    "plt.legend()\n",
    "\n",
    "plt.tight_layout()\n",
    "plt.show()"
 ]
}
```
